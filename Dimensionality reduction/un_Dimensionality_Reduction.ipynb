```python
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
digits = load_digits()

X = digits.data
y = digits.target

print(X.shape)
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

print(X_pca.shape)
```

```python
plt.figure(figsize=(10, 7))
scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=y,
    alpha=0.7
)

cbar = plt.colorbar(scatter)
cbar.set_label('Digit Label')

plt.title("PCA Visualization of Digits Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(
    pca.explained_variance_ratio_,
    marker='o'
)

plt.title("Explained Variance by Principal Components")
plt.xlabel("Component Index")
plt.ylabel("Variance Ratio")
plt.show()
```

* Used the Digits dataset from scikit-learn
* Applied PCA (Principal Component Analysis) for dimensionality reduction
* Reduced dataset features from 64 dimensions to 2 dimensions
* Extracted the most important patterns and variance from the data
* Visualized the transformed data using a scatter plot
* Observed clustering of similar handwritten digits (0–9)
* Learned how unsupervised learning can simplify high-dimensional data
* Understood the importance of PCA in visualization and preprocessing
* Explored explained variance to measure retained information
* Improved understanding of feature extraction and data compression
